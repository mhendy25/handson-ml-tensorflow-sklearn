- broadcasting 
- rule 1: numpy compares the shapes of the two arrays from right to left, if two dims aren't the same: if any is 1, it's fine, otherwise, it's a value error

```python
h = np.arange(5).reshape((1, 1, 5))
h
```

```python
h + [1, 2, 3, 4, 5]
```

- rule 2: if the operation is compatible and an array has 1 along a dimension, then the final dimension is the max of both curr dims, and the value at dim 1 is repated along that axis 

```python
k = np.arange(6).reshape(2, 3)
k
```

```python
k + [[100], [200]] # (2, 3) + (2 * 1) = (2, 3)
```

```python
k + [100, 200, 300] # (2, 3) + (3,)
```

```python
k + 1000
```
